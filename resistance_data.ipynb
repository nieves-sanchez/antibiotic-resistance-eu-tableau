```python
import os
import pandas as pd
import numpy as np

files = os.listdir("data/raw")

dfs = []
for file in files:
    if file.endswith(".csv"):
        dfs.append(pd.read_csv(f"data/raw/{file}"))

df = pd.concat(dfs, ignore_index=True)
```

```python
df[["Bacteria", "AntibioticGroup"]] = df["Population"].str.split("|", expand=True)
```

```python
df.drop(columns=["HealthTopic", "Population", "Unit", "Indicator", "TxtValue"], inplace=True)
```

```python
df.rename(columns={"Time": "Year", "RegionCode": "CountryCode", "RegionName": "Country", "NumValue": "Resistance_Percentage"}, inplace=True)
```

```python
df = df[["Bacteria", "AntibioticGroup", "Year", "CountryCode", "Country", "Resistance_Percentage"]]
```

```python
df.head()
```

```python
def eda_amr(df):
    """
    Analisis exploratorio completo para dataset de resistencia antimicrobiana (AMR)
    """
    
    print("🔎 PRIMER VISTAZO A LOS DATOS")
    print("Primeras filas:")
    display(df.head())
    
    print("\nUltimas filas:")
    display(df.tail())
    
    print("\nMuestra aleatoria:")
    display(df.sample(5))
    
    
    print("\n📐 ESTRUCTURA DEL DATAFRAME")
    print(f"Dimensiones (filas, columnas): {df.shape}\n")
    
    print("Columnas:")
    display(df.columns)
    
    print("\nInformacion general:")
    df.info()
    
    print("\nTipos de datos por columna:")
    display(df.dtypes)
    
    
    print("\n📊 ESTADISTICAS DESCRIPTIVAS")
    print("Variables numericas:")
    display(df.describe().T)
    
    print("\nVariables categoricas:")
    display(df.describe(include='object').T)
    
    
    print("\n🔢 CARDINALIDAD")
    print("Numero de valores unicos por columna:")
    display(df.nunique())
    
    
    print("\n🧩 VALORES UNICOS CLAVE")
    
    print("\nBacterias:")
    display(df["Bacteria"].unique())
    
    print("\nGrupos antibioticos:")
    display(df["AntibioticGroup"].unique())
    
    print("\nAños:")
    display(sorted(df["Year"].unique()))
    
    print("\nPaises (primeros 20):")
    display(df["Country"].unique()[:20])
    
    
    print("\n🚨 VALORES NULOS")
    display(df.isnull().sum())
    
    print("\nPorcentaje de valores nulos (%)")
    display(((df.isnull().sum() / df.shape[0]) * 100).round(2))
    
    
    print("\n🔁 DUPLICADOS")
    dup_count = df.duplicated(
        subset=["Bacteria","AntibioticGroup","Year","CountryCode"]
    ).sum()
    
    print(f"Duplicados segun combinacion clave: {dup_count}")
    
    if dup_count > 0:
        display(df[df.duplicated(
            subset=["Bacteria","AntibioticGroup","Year","CountryCode"],
            keep=False
        )])
    else:
        print("No hay duplicados estructurales.")   
    
    print("\n📈 DISTRIBUCION RESISTANCE_PERCENTAGE")
    display(df["Resistance_Percentage"].describe()
    )
```

```python
eda_amr(df)
```

```python
df["Resistance_Percentage"] = (
    df["Resistance_Percentage"]
    .replace("-", np.nan)
)
```
